## What is python?

Python was created by [Guido van Rossum](https://gvanrossum.github.io/) and first released in 1991, designed to be simple, readable, and easy to learn. Over the decades it has grown into one of the world's most popular languages, powering everything from web development and automation to data science and artificial intelligence. Its value lies in its clean syntax, vast library ecosystem, and strong community, which make it both beginner-friendly and capable enough for large-scale professional work.

## Pros and Cons

### Pros

- **Readable and concise** – Clean syntax lets you express ideas in fewer lines, speeding up development and making code easier to maintain.
- **Huge ecosystem** – A vast collection of third-party libraries (`NumPy`, `Django`, `TensorFlow`, etc.) covers web, data, AI, and automation needs.

- **Weak in mobile and some concurrency cases** – It's not a first choice for mobile development, and the GIL limits true parallel threading for CPU-bound tasks.

## Variables and values

A **value** is a piece of data: a number, a string, a list. A **variable** is a *name* that refers to a value. Think of a variable as a label you attach to a value, not as a box the value lives in.

```python
murphy = "The sun shone, having no alternative, on the nothing new"
```

Here `"The sun shone..."` is the value, and `murphy` is the name we've given it.

* **The `=` sign means "make this name refer to this value."** It is not the `=` from maths. It is an instruction, not a statement that two things are equal.

* **Python works out the right-hand side first, then attaches the name.**
```python
    total = 3 + 4    # Python calculates 7, then attaches the name total to it
```

* **A name can be moved to a different value.** This is called reassignment:
```python
    score = 10
    score = 25       # score now refers to 25; the 10 is forgotten
```

* **This is why `x = x + 1` isn't nonsense.** In maths it would be impossible. In Python it means: take the value `x` currently refers to, add 1, then move the name `x` to the result.
```python
    x = 5
    x = x + 1        # x is now 6
```

* **Using a name gives you its value.** When you write `len(murphy)`, Python looks up what `murphy` refers to and passes that string to `len`.

* **Names are case-sensitive, and a name that was never assigned doesn't exist.**
```python
    Murphy           # NameError: name 'Murphy' is not defined
```

* **Two names can refer to the *same* value.** With lists, which can be changed, this matters:
```python
    a = ["apple", "pear"]
    b = a            # b is a second label on the same list, not a copy
    b.append("fig")
    print(a)         # ['apple', 'pear', 'fig'], so a changed too!
```
    This happens because there is only one list, with two labels on it.

## Lists and Strings

Today we are going to learn about two of the most basic datatypes in python: `lists` and `strings`. There are many others, but these give us a useful starting point. 

* A list is an ordered, changeable collection of items, written with square brackets. Items can be any type and can repeat.
  
    `sample_list = ["apple", 3.14, "Caesar Augustus", 42, "sputnik", False, "apple", [['angiogram', 17], ['fromage']]]`
  
* A string is an ordered, unchangeable sequence of characters, written with quotes. You can't modify it in place, but you can create new strings from it.

    `murphy = "The sun shone, having no alternative, on the nothing new."`

### Indexing

In python, counting starts at `0`. This means that the first item known as the `zeroeth` item. Indexing an item means identifying it based on its position in a sequence. These are the basic principles of indexing:

* To count the number of items in a sequence `s`, use `len(s)`.

* You access a single element of a sequence `s` with `s[i]`, where `i` is an integer position. Indexing starts at `0` (the first element), and negative indices count from the end (`-1` is the last element). Out-of-range indices raise an `IndexError`.

* If a sequence contains a subsequence (like a `list` containing a `list`, concatenate the indexes `s[i][j]`, where `i` ia the position of the subsequence, and `j` is the position of the item in the subsequence.

A `slice` takes a range of items from a sequence. The rule of slicing is to start with the first item you want and stop at the first item you *don't* want. If you want to count in steps of more than `1`, specify the number of steps in between the items. The format is `s[start:stop:step]`.


#### List indexing

````python
sample_list = ["apple", 3.14, "Caesar Augustus", 42, "sputnik", False, "apple", [['angiogram', 4-5j], ['fromage']]]

print(len(sample_list))
print(sample_list[0])
print(sample_list[1])
print(sample_list[-1])
print(sample_list[7][1])
print(sample_list[7][0][0])
print(sample_list[0:6:2])
````

#### String indexing

````python
murphy = "The sun shone, having no alternative, on the nothing new."

print(len(murphy))
print(murphy[0])
print(murphy[1])
print(murphy[-1])
print(murphy[0:13])
print(murphy[::-1]) #Reverse steps go from right to left, changing the left to right default
print(murphy[0:56:3])
````

# Making python useful: The pubs of Britain

Like all programming languages, python is only as valuable as the things it can do. In this part of the class, we're going to explore how we can use some geo and language data to look at how pub names are distributed across the UK

The data source is the Food Safety Authority list of licensed premises, an archived version of which can be found at [Kaggle.com](https://www.kaggle.com/datasets/getthedata/open-pubs). This gives us lots of interesting data to work with.

### Import libraries and data

The next cell imports some libraries we'll be using. These are specialist packages in python that perform specific tasks. You don't need to worry about the details of packages for now. It also imports the data from the `.csv` format. This stands for 'comma separated variables' and is a universal data format. It is very like (and entirely compatible with) the `.xlsx` Excel spreadsheet format.

````python
import re # a library for manipulating string data
from collections import Counter # a library for counting collections of data in sophisticated ways
import pandas as pd # an extremely valuable library! Pandas is a data science library, the 'Excel' of python
import matplotlib.pyplot as plt # A plotting library
import seaborn as sns # another, simpler plotting library
sns.set() # sets seaborn's style

df = pd.read_csv("open_pubs.csv") # opens our data as a pandfas dataframe (df)

df = df[df.northing > 0]  # drop 63 rows with missing coordinates

df
````

### Process the data so it only includes pubs, then find the most common word pairings

````python

# Keep pubs and inns: drop clubs, hotels, bars etc.
not_pubs = r"\bclub\b|legion|social|golf|cricket|bowling|rugby|football|sports|association|ltd|conservative|labour|working men|hotel|restaurant|\bbar\b|cafe|kitchen"
pubs = df[~df.name.str.contains(not_pubs, case=False)]

# We want to remove these words because they don't give us much information. This is descision; you might choose differently!
stopwords = {"the", "and", "of", "at", "in", "on", "a", "an", "to", "by", "ye", "olde"}
pub_words = {"inn", "arms", "house", "public", "tavern", "ph", "pub"}  # generic words for a pub

# This code counts the most commonly occuring pairs of words and creates a table
counts = Counter()
for name in pubs.name:  # pubs = the filtered DataFrame from the map cell
    name = name.lower().replace("'", "").replace("’", "")  # King's -> kings, not king + s
    name = re.sub(r"[^a-z\s]", " ", name)                  # other punctuation and digits -> space
    words = [w for w in name.split()
             if w not in stopwords and w not in pub_words and len(w) > 1]
    counts.update(zip(words, words[1:]))

table = pd.DataFrame(counts.most_common(30), columns=["word", "count"])
table
````

## Where are the pubs located?

Our data contains location details for each pub, both in terms of latitude and longitude and eastings and northings. Eastings and northings are easier to deal with, so we're going to use them. But what are they?

The Ordnance Survey divides Britain in a set of grids that cover every area in mainland Britain (Northern Ireland is part of the Irish grid.) This grid is as per the graphic below:

![Grid](Uk_grid.jpg)

The origin of the grid is located southwest of the Isles of Scilly. Eastings and northings are six-digit numbers that tell you how far away from the origin a place is.

* The first three digits tell you how many <i>kilometres</i> east or north of the origin the location is. The convention is easting first, then northing.
* The last three digits tell you how far east or north you are from the first location in <i>metres</i>.
* Colchester is represented by `604748` `234405`. This means it is 604km and 748m east of the origin, and 234km and 405m north of it.

````python
fig, ax = plt.subplots(figsize=(8, 8))
ax.scatter(pubs.easting, pubs.northing, s=0.3, c="black", alpha=0.4, linewidths=0)

ax.set_aspect("equal")  # eastings and northings are both metres, so the map isn't distorted
ax.axis("off")
plt.title('British pubs by location')
plt.show()
````

## What about other common pub names?

Though British pubs have lots of nautical associations, these are far from exhaustive. Agricultural and arboreal terms are very common, as are references to rural activities like hunting and fishing. Can we perhaps see how these names map onto the types of land the pubs are near? The ESA WorldCover land cover data allows us to do this (© ESA WorldCover project 2021, used under CC BY 4.0). 

````python
land = pd.read_csv("landcover.csv")

print(land.type.value_counts())
land.head()
````

These are squares rather than lines, so we're back to `ax.scatter` — exactly what we used
for the pubs. `marker="s"` draws each one as a little square instead of a dot, so they tile
together into regions.

````python
fig, ax = plt.subplots(figsize=(8, 12))

for kind, colour in [("arable", "#e8d79a"), 
                     ("forest", "#2d6a3e"), 
                     #("urban", "#8e7cc3")
                    ]:
    subset = land[land.type == kind]
    ax.scatter(subset.easting, subset.northing, s=1.2, c=colour,
               marker="s", linewidths=0, label=kind)

term1 = "plough"
term2 = "woodman"
term3 = "oak"

# Overlay the week-1 hypotheses
for word, colour in [(term1, "tab:orange"), 
                     (term2, "tab:blue"),
                         (term3, "tab:red")
                    ]:
    subset = pubs[pubs.name.str.contains(word, case=False)]
    ax.scatter(subset.easting, subset.northing, s=6, c=colour, label=word)


ax.scatter(pubs.easting, pubs.northing, s=0.2, c="black", alpha=0.4, linewidths=0)

ax.set_aspect("equal")
ax.axis("off")
ax.legend(markerscale=6)
plt.title('Britsh pub name words by type of land')
plt.show()
````

### Does the landscape show up in the pub names?

Earlier we counted the words in pub names. Now we can ask whether those words know where
they are. Each pub sits in one of these 1 km squares, so we can look up the land cover for
every pub and compare.

````python
# Which square is each pub in? Round its position down to the nearest kilometre.
located = pubs.assign(
    cell_easting=(pubs.easting // 1000) * 1000 + 500,
    cell_northing=(pubs.northing // 1000) * 1000 + 500,
).merge(land, left_on=["cell_easting", "cell_northing"],
        right_on=["easting", "northing"], suffixes=("", "_cell"))

print(f"{len(located):,} pubs matched to a square\n")

everywhere = located.type.value_counts(normalize=True)

rows = []

words = ["plough", "woodman", "oak", "shepherd", "miners", "fisherman"]
for word in words:
    subset = located[located.name.str.contains(word, case=False)]
    share = subset.type.value_counts(normalize=True)
    rows.append({"word": word,
                 "pubs": len(subset),
                 "in arable %": round(share.get("arable", 0) * 100, 1),
                 "in grassland %": round(share.get("grassland", 0) * 100, 1),
                 "in wetland %": round(share.get("wetland", 0) * 100, 1),
                 "in urban %": round(share.get("urban", 0) * 100, 1),
                 "in water %": round(share.get("water", 0) * 100, 1),
                 "in bare %": round(share.get("bare", 0) * 100, 1),
                 "in shrub %": round(share.get("shrub", 0) * 100, 1),
                 "in forest %": round(share.get("forest", 0) * 100, 1)})

print(f"any pub:  arable {everywhere['arable'] * 100:.1f}%   "
      f"forest {everywhere['forest'] * 100:.1f}%")
pub_table = pd.DataFrame(rows)
pub_table
````

````python
from scipy.stats import chi2_contingency

land_cols = ["in arable %", "in grassland %", "in urban %", "in water %", "in forest %"]

# rebuild counts
counts = pub_table[land_cols].apply(
    lambda col: (col / 100 * pub_table["pubs"]).round().astype(int)
)
counts.index = pub_table["word"]   # label rows by word

df_counts = pd.DataFrame(counts)

print(df_counts)

chi2, p, dof, expected = chi2_contingency(counts)
print("\n", f"chi2 = {chi2:.2f}, p = {p:.4f}, dof = {dof}")
````
