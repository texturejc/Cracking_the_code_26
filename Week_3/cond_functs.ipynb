# Cracking the Code Week 3

This week we are going to introduce two new concepts: the `conditional` and the `function`. `Conditionals` allow us to tell our code to do specific things when specific conditions are met, and hence play an important role in making code responsive. `Functions` allow us to parcel complex code into a single, easy-to-use package that saves having to rewrite it every time we want to use it.


# Conditionals

A **conditional** lets your program choose between different paths depending on whether a test is `True` or `False`. Without conditionals, code runs top to bottom, the same way every time. With them, the program can **branch**: do one thing in one situation, another thing in a different situation. The test itself is any expression that evaluates to a **truthy** or **falsy** value — commonly a comparison like `x > 5`, but also just a variable, a container, or `None`.

To write a conditional in Python, you use the keyword `if`, a test expression, and a colon `:`. The body is indented beneath it:

```python
    temperature = 30

    if temperature > 25:
        print("It's warm today")
```

The indented block only runs when the test is `True`. When it's `False`, Python skips the block entirely and moves on.

## `else` — the fallback path

Use `else` to say "otherwise, do this instead". It runs when the `if` test is `False`:

```python
    temperature = 10

    if temperature > 25:
        print("It's warm today")
    else:
        print("It's not warm today")
```

Exactly one of the two branches runs — never both, never neither.


## `elif` — more than two options

When you have several possible cases, chain them with `elif` ("else if"). Python tests each condition **in order** and runs the **first** one that's `True`, then skips the rest:

```python
    temperature = 15

    if temperature > 25:
        print("Hot")
    elif temperature > 15:
        print("Warm")
    elif temperature > 5:
        print("Cool")
    else:
        print("Cold")
```

Note that the **order matters**. If you put `temperature > 5` first, a temperature of `20` would match that branch and never reach `> 15`. Put the most specific (or most extreme) tests first, and the broadest last.


## Comparisons you can test

Inside the test you can use the usual comparison operators, and combine them with `and`, `or`, and `not`:

```python
    age = 25

    if age >= 18 and age < 65:
        print("Working age")

    if age < 18 or age >= 65:
        print("Not working age")

    if not age == 30:
        print("Not 30")
```

| Operator | Meaning |
|----------|---------|
| `==` | equal to |
| `!=` | not equal to |
| `<` `<=` | less than / less than or equal |
| `>` `>=` | greater than / greater than or equal |
| `and` `or` `not` | combine or negate tests |

A common pitfall: `=` assigns, `==` compares. `if x = 5:` is a **syntax error**; you want `if x == 5:`.

## Truthiness — tests don't have to be comparisons

Any value can be used as a test. Python treats certain values as **falsy** and everything else as **truthy**:

```python
    # These are all falsy:
    #   False, None, 0, 0.0, "", [], {}, set()

    name = ""

    if name:
        print("Hello, " + name)
    else:
        print("No name given")
```

This is idiomatic Python: `if name:` is preferred over `if len(name) > 0:` or `if name != "":`. It works for any container — an empty list, dict, or string is falsy, a non-empty one is truthy:

```python
    postcodes = []

    if postcodes:
        print(f"{len(postcodes)} postcodes on file")
    else:
        print("No postcodes yet")
```

## Conditionals inside loops

Conditionals and `for` loops are frequently combined — the loop does the repetition, the conditional does the filtering:

```python
    numbers = [3, 7, 2, 9, 4]

    evens = []
    for n in numbers:
        if n % 2 == 0:
            evens.append(n)
    print(evens)          # [2, 4]

    for n in numbers:
        if n > 5:
            print(n, "is big")
        else:
            print(n, "is small")
```

The conditional is re-evaluated on every pass of the loop, so a different branch can run each time.


## Conditionals inside conditionals

A conditional can contain another conditional — this is called **nesting**:

```python
    temperature = 15

    if temperature > 5:
        if temperature > 25:
            print("Hot")
        else:
            print("Mild")
    else:
        print("Cold")
```

The inner `if` only runs when the outer `if` is `True`. Note that the `elif` chain from earlier is usually cleaner than nesting — prefer it when the cases are mutually exclusive:

```python
    # Same logic, flatter and easier to read:
    if temperature > 25:
        print("Hot")
    elif temperature > 5:
        print("Mild")
    else:
        print("Cold")
```

Nest when the inner test only makes sense in the context of the outer one (e.g. checking a key exists before checking its value).


## Blending conditionals with containers

Because a `dict` can hold `list`s and other `dict`s, conditionals are useful for checking whether a key or value is present before acting on it:

```python
    animals = {'dog': 'mammal', 'spider': 'arachnid', 'hornet': 'insect', 'toad': 'amphibian'}

    if 'dog' in animals:
        print("dog is", animals['dog'])      # dog is mammal

    if 'cat' not in animals:
        print("No cat on file")              # No cat on file
```

The `in` operator works for any container — `dict` (checks keys), `list` (checks values), `str` (checks substrings), and `set` — and returns a boolean you can test directly.

```python
    postcodes = ["EH1", "EH8", "EH16"]

    if "EH8" in postcodes:
        print("EH8 is covered")
```

Note that for a `dict`, `in` checks **keys**, not values. `'mammal' in animals` is `False`, because `'mammal'` is a value, not a key.

## Summary

| Keyword | Runs when |
|---------|-----------|
| `if` | the test is `True` |
| `elif` | the previous tests were `False` and this test is `True` |
| `else` | all previous tests were `False` |

A few things to remember:

- Exactly **one** branch of an `if`/`elif`/`else` chain runs.
- `elif` tests are checked **top to bottom**; the first match wins.
- Any value can be a test — empty containers, `0`, `None`, and `""` are **falsy**.
- Use `==` to compare, `=` to assign.
- Conditionals combine naturally with `for` loops for filtering, and with `in` to check for membership.

# Functions

A **function** is a named, reusable block of code that performs a specific task. Instead of writing the same lines over and over, you write them once inside a function and **call** it whenever you need it. Functions take **inputs** (called **arguments**), do some work, and optionally **return** an output. They are the main tool for organising a program: they let you break a big problem into small, named, testable pieces, and they keep your code **DRY** ("Don't Repeat Yourself").

To define a function in python, you use the keyword `def`, a name, a pair of parentheses containing any **parameters**, and a colon `:`. The body is indented beneath it:

```python
    def greet():
        print("Hello, Edinburgh")

    greet()          # Hello, Edinburgh
    greet()          # call it again — the code runs again
```

The `def` line only *defines* the function; nothing runs until you **call** it by name with `()`. Note that omitting the parentheses (`greet` instead of `greet()`) refers to the function object itself without running it — a common beginner bug.

## Parameters — giving a function inputs

A function becomes far more useful when it can accept values. The names in the parentheses are **parameters**; the values you pass in when calling are **arguments**:

```python
    def greet(city):
        print("Hello, " + city)

    greet("Glasgow")     # Hello, Glasgow
    greet("Aberdeen")    # Hello, Aberdeen
```

The parameter `city` acts like a variable that is set to whatever argument you pass. It only exists inside the function.


## `return` — giving a function an output

`print` displays a value on screen; `return` hands a value **back** to the caller so it can be stored, reused, or combined. These are not the same thing:

```python
    def double(n):
        return n * 2

    result = double(5)
    print(result)         # 10
```

Once a `return` runs, the function stops immediately. Code after it is not executed:

```python
    def double(n):
        return n * 2
        print("never runs")   # unreachable

    print(double(5))      # 10
```

A function with no `return` (or a bare `return`) gives back `None`:

```python
    def show(n):
        print(n)          # prints, but returns None

    x = show(5)           # prints 5
    print(x)              # None
```

`print` is for the human reading the output; `return` is for the program to use the value. Use `return` when the result is needed elsewhere.


## Reusing the earlier examples

Functions let you wrap up the patterns from previous sections so you can call them on different data. Here is the postcode lookup from the conditionals section, turned into a function:

```python
    postcodes = {"EH1": "Edinburgh", "G1": "Glasgow", "M1": "Manchester"}

    def lookup(postcode):
        if postcode in postcodes:
            return postcodes[postcode]
        else:
            return "Postcode not found"

    print(lookup("G1"))      # Glasgow
    print(lookup("ZZ9"))     # Postcode not found
```

And here is the filtering loop from the iteration section, wrapped up:

```python
    animals = {'dog': 'mammal', 'spider': 'arachnid', 'hornet': 'insect',
               'toad': 'amphibian', 'cat': 'mammal', 'beetle': 'insect'}

    def animals_of_type(kind):
        result = []
        for name, animal_type in animals.items():
            if animal_type == kind:
                result.append(name)
        return result

    print(animals_of_type("mammal"))    # ['dog', 'cat']
    print(animals_of_type("insect"))    # ['hornet', 'beetle']
```

Notice how the function captures a whole pattern — loop, conditional, accumulator, return — behind a single name. Once defined, you can call it with any `kind` without rewriting the logic.

## Parameters with defaults

You can give a parameter a **default value** so the caller can omit it:

```python
    def greet(city, greeting="Hello"):
        return f"{greeting}, {city}"

    print(greet("Glasgow"))                  # Hello, Glasgow
    print(greet("Glasgow", "Good morning"))  # Good morning, Glasgow
```

Parameters with defaults must come **after** those without. A caller can also name arguments explicitly, which is clearer when there are several:

```python
    print(greet(city="Glasgow", greeting="Hi"))   # Hi, Glasgow
```


## Multiple parameters

A function can take as many parameters as you like, separated by commas:

```python
    def describe(city, population, country="Scotland"):
        return f"{city} ({country}) has a population of {population}"

    print(describe("Edinburgh", 506520))
    # Edinburgh (Scotland) has a population of 506520

    print(describe("Cardiff", 366000, "Wales"))
    # Cardiff (Wales) has a population of 366000
```


## Scope — variables inside a function are private

A variable created inside a function only exists there. It vanishes when the function returns:

```python
    def double(n):
        result = n * 2      # 'result' lives only inside double
        return result

    double(5)
    print(result)           # NameError: name 'result' is not defined
```

This is a feature, not a bug: functions don't accidentally clobber the caller's variables. To get a value out, **return** it.

A function *can* read variables defined outside it (like the `postcodes` and `animals` dicts above), but it's usually clearer to pass what it needs as a parameter. Prefer parameters over relying on outside variables — it makes the function self-contained and easier to test.


## Why functions matter

| Benefit | What it means in practice |
|---------|---------------------------|
| **Reuse** | Write the logic once, call it many times |
| **Naming** | `animals_of_type("mammal")` says what it does |
| **Testing** | Small functions are easy to check in isolation |
| **Organisation** | Break a big problem into small, named steps |
| **Avoiding repetition** | Fix a bug once, everywhere benefits |

## Summary

- Define a function with `def name(parameters):`; call it with `name(arguments)`.
- **Parameters** are the inputs in the definition; **arguments** are the values you pass in.
- **`return`** sends a value back to the caller; **`print`** just displays it. They are not interchangeable.
- A function with no `return` gives back `None`.
- Variables created inside a function are **local** — they don't exist outside it.
- Give parameters **default values** to make them optional.
- Small functions that call other small functions are the foundation of readable programs.

# Practical demo: a headline-driven chip trader

Today we'll use conditionals and functions to build a **toy trading algorithm**. The idea is simple:

> Pick some words. If those words appear in yesterday's news headlines, make a trade today.

We'll test it on real data: **semiconductor (computer chip) stocks** such as Nvidia, AMD and Intel, and about 250,000 real news headlines from **2023 to 2025**.

The plan:

1. Load the data (two JSON files, just like last week).
2. Write a function that works out how much a stock moved in a day.
3. Write a function that counts how often our words appear in a day's headlines.
4. Write a function that uses **conditionals** to decide what to do: buy, sell, or sit out.
5. Write a function that runs our rules over every day in the data.
6. Write a function that scores the result with the **Sharpe ratio**.

Each step is a small function. At the end we plug them together.

*This is a toy for learning Python. It is not investment advice!*

## Step 0: Load the data

There are two files:

- `headlines.json` is a **dict** where each key is a date and each value is a **list** of that day's headlines. The headlines have been lower-cased and had their punctuation removed, to make word matching easier.
- `semis_prices.json` is a **dict of dicts**. The outer key is the stock's ticker symbol (e.g. `"NVDA"` for Nvidia), and the inner dict maps each date to that day's **closing price** in US dollars.

Dates are written as text in the form `"YYYY-MM-DD"`.

````python
import json

with open("headlines.json", "r", encoding="utf-8") as f:
    headlines = json.load(f)

with open("semis_prices.json", "r", encoding="utf-8") as f:
    prices = json.load(f)
````

````python
print(list(prices.keys()))             # the stocks we have

nvidia = prices["NVDA"]
print(len(nvidia), "days of prices")
print(nvidia["2024-03-01"])            # Nvidia's closing price on 1 March 2024
````

## Step 1: How much did the stock move?

A **return** is the percentage change in price from one day to the next. If a stock goes from £100 to £103, the return is $(103 - 100) / 100 = 0.03$, which is 3%.

We'll need this calculation hundreds of times, so we wrap it in a function.

````python
def daily_return(price_today, price_yesterday):
    return (price_today - price_yesterday) / price_yesterday


print(daily_return(103, 100))    # 0.03, a 3% rise
print(daily_return(95, 100))     # -0.05, a 5% fall
````

## Step 2: Do the headlines mention our words?

We give the function a list of headlines and a list of words. It goes through every headline, splits it into words with `.split()`, and counts every time one of our words shows up.

Why split into words instead of checking `word in headline` directly? Because `in` on a string checks for **any matching piece of text**, so `"ai" in "said"` is `True`. Splitting first means we only match whole words.

````python
def count_mentions(day_headlines, words):
    count = 0
    for headline in day_headlines:
        headline_words = headline.split()
        for word in words:
            if word in headline_words:
                count = count + 1
    return count


test_headlines = ["nvidia shares hit record high", "chip stocks fall on china fears", "rain expected in edinburgh"]

print(count_mentions(test_headlines, ["nvidia", "record"]))   # 2: both words are in the first headline
print(count_mentions(test_headlines, ["china"]))              # 1
print(count_mentions(test_headlines, ["tariff"]))             # 0
````

## Step 3: Decide what to do (conditionals!)

We'll have two lists of words:

- **buy words**: words we think are good news for chip stocks, like `"record"` or `"ai"`.
- **sell words**: words we think are bad news, like `"tariffs"` or `"recession"`.

Each day we count both and compare them. This is our **position**, a number that says what we're holding:

| Situation | Position | What it means |
|-----------|----------|---------------|
| More buy words than sell words | `1` | **Buy**: we make money if the stock goes **up** |
| More sell words than buy words | `-1` | **Sell short**: a bet that the stock goes **down**, so we make money if it falls |
| A tie (including no mentions at all) | `0` | **Sit out**: hold cash, make nothing, lose nothing |

This is exactly an `if` / `elif` / `else` chain: three cases, and exactly one of them must happen.

````python
def decide_position(buy_count, sell_count):
    if buy_count > sell_count:
        return 1       # buy
    elif sell_count > buy_count:
        return -1      # sell short
    else:
        return 0       # sit out


print(decide_position(5, 2))    # 1
print(decide_position(0, 3))    # -1
print(decide_position(4, 4))    # 0
````

The position makes the maths very neat. Our return for the day is just **position × the stock's return**:

- Position `1` and the stock rises 2%: we make `1 × 0.02 = 0.02` (+2%)
- Position `-1` and the stock rises 2%: we make `-1 × 0.02 = -0.02` (−2%, our bet was wrong)
- Position `-1` and the stock falls 2%: we make `-1 × -0.02 = 0.02` (+2%, our bet was right)
- Position `0`: we make `0`, whatever happens

## Step 4: Run the rules over every day

This is called a **backtest**: we pretend we followed our rules in the past, and see how we would have done.

For each trading day we:

1. look up **yesterday's** headlines (we can only trade on news we've already seen!),
2. count the buy words and sell words,
3. decide our position,
4. work out the stock's return **today**, and multiply it by our position.

We loop through the dates using their position in the list (`i`), so that `dates[i - 1]` is the trading day before `dates[i]`. We start at `i = 1` because the very first day has no day before it.

As well as our strategy's returns, we also record the plain **buy-and-hold** return: what we'd get by simply owning the stock every day and ignoring the news. That gives us something to compare against.

This function **returns two lists**. Python lets you return several values separated by commas, and you catch them the same way: `a, b = my_function()`.

````python
def backtest(ticker, buy_words, sell_words):
    stock = prices[ticker]
    dates = sorted(stock)    # all the trading days, in date order

    strategy_returns = []
    buy_and_hold_returns = []

    for i in range(1, len(dates)):
        today = dates[i]
        yesterday = dates[i - 1]

        # 1. Yesterday's news (with an empty list as a fallback, in case a day is missing)
        if yesterday in headlines:
            news = headlines[yesterday]
        else:
            news = []

        # 2. and 3. Count the words and decide what to do
        buy_count = count_mentions(news, buy_words)
        sell_count = count_mentions(news, sell_words)
        position = decide_position(buy_count, sell_count)

        # 4. How did the stock do today, and how did we do?
        stock_return = daily_return(stock[today], stock[yesterday])
        strategy_returns.append(position * stock_return)
        buy_and_hold_returns.append(stock_return)

    return strategy_returns, buy_and_hold_returns
````

## Step 5: Was it any good? The Sharpe ratio

The obvious way to score a strategy is "how much money did it make?" But that only tells half the story. Making 10% a year by taking wild gambles is not as impressive as making 10% a year smoothly and steadily, because the gambler could just as easily have lost a fortune.

The **Sharpe ratio** (named after economist William Sharpe) measures **return per unit of risk**:

$$\text{Sharpe ratio} = \frac{\text{average return}}{\text{standard deviation of returns}}$$

- The **average return** is how much you make on a typical day.
- The **standard deviation** measures how **bumpy** the ride is: how far each day's return tends to be from the average. Bigger swings mean a bigger standard deviation, which means more risk.

$$\sigma = \sqrt{\frac{1}{N}\sum_{i=1}^{N}(x_i - \mu)^2}
$$

So a high Sharpe ration means you're making a strong, steady return on your investment

| **Smug Rees** | 1 | 1 | 2 | 1 | 0 | **1** |
| **Reeval Rachel** | 6 | −4 | 5 | −3 | 1 | **1** |

They have the same average return: 1% a day. But Reeval Rachel's ride is far bumpier that Smug Rees's. Let's work out the standard deviation by hand:

1. Subtract the average from each day, and square the result. Squaring makes every value positive, so ups and downs both count as "bumpiness".
   - Smug Rees: $0^2, 0^2, 1^2, 0^2, (-1)^2$ = 0, 0, 1, 0, 1
   - Reeval Rachel: $5^2, (-5)^2, 4^2, (-4)^2, 0^2$ = 25, 25, 16, 16, 0
2. Take the average of those squares:
   - Smug Rees: $2 / 5 = 0.4$
   - Reeval Rachel: $82 / 5 = 16.4$
3. Take the square root, which undoes the squaring:
   - Smug Rees: $\sqrt{0.4} \approx 0.63$
   - Reeval Rachel: $\sqrt{16.4} \approx 4.05$

Now the Sharpe ratios:

- Smug Rees: $1 / 0.63 \approx 1.58$
- Reeval Rachel: $1 / 4.05 \approx 0.25$

Smug Rees and Reeval Rachel have the same average return, but Smug Rees's Sharpe ratio is over **six times** Reeval Rachel's. Smug Rees got the same reward for far less risk.

### Making it yearly

Sharpe ratios are normally quoted **per year**. The convention is to multiply the daily ratio by $\sqrt{252}$ (about 15.9), because there are about 252 trading days in a year. You don't need to worry about why it's the square root: just know that it's the standard scaling, so everyone's numbers can be compared fairly.

A rough guide for the **yearly** Sharpe ratio:

| Sharpe ratio | Verdict |
|--------------|---------|
| Below 0 | Lost money |
| 0 to 0.5 | Not much reward for the risk |
| 0.5 to 1 | Decent |
| 1 to 2 | Good |
| Above 2 | Excellent (and if it's from a backtest, be suspicious!) |

*(Professionals also subtract the "risk-free" interest rate you could get from a savings account. We'll leave that out to keep things simple.)*

Here's the calculation as functions. Notice how `sharpe_ratio` calls the two smaller functions: small functions building on each other.

````python
def average(numbers):
    return sum(numbers) / len(numbers)


def standard_deviation(numbers):
    avg = average(numbers)
    total = 0
    for n in numbers:
        total = total + (n - avg) ** 2      # squared distance from the average
    return (total / len(numbers)) ** 0.5    # ** 0.5 is the square root


def sharpe_ratio(daily_returns, yearly=True):
    sd = standard_deviation(daily_returns)
    if sd == 0:
        return 0                             # never traded: avoid dividing by zero
    ratio = average(daily_returns) / sd
    if yearly:
        ratio = ratio * 252 ** 0.5
    return ratio
````

````python
Smug_Rees = [1, 1, 2, 1, 0]
Reeval_Rachel = [6, -4, 5, -3, 1]

print("Smug Rees:  ", average(Smug_Rees), round(standard_deviation(Smug_Rees), 2), round(sharpe_ratio(Smug_Rees, yearly=False), 2))
print("Reeval Rachel:", average(Reeval_Rachel), round(standard_deviation(Reeval_Rachel), 2), round(sharpe_ratio(Reeval_Rachel, yearly=False), 2))
````

These match the numbers we worked out by hand.

### How did the money grow?

A picture helps. This function starts with £1 and applies each day's return in turn, keeping a record of how much we have after every day. A return of 0.02 turns £1.00 into £1.02, so each day we multiply by `(1 + return)`.

````python
import matplotlib.pyplot as plt


def growth_of_one_pound(daily_returns):
    money = 1
    history = []
    for r in daily_returns:
        money = money * (1 + r)
        history.append(money)
    return history


plt.figure(figsize=(10, 5))
plt.plot(growth_of_one_pound(buy_and_hold), label="Buy and hold", color="grey")
plt.plot(growth_of_one_pound(strategy), label="Headline strategy", color="tab:blue")
plt.xlabel("Trading days since January 2023")
plt.ylabel("Value of £1 invested")
plt.title(f"{ticker}: headline strategy vs buy and hold")
plt.legend()
plt.show()
````

## A word of warning

If you keep tweaking your words until the Sharpe ratio looks great, you've almost certainly just found words that happened to work **in this particular past**. This is called **overfitting**, and it's the most common way trading strategies fool their creators. A few other things this toy ignores:

- **Trading costs**: real trades cost money, and this strategy trades a lot.
- **Timing**: some of "yesterday's" headlines were published after the market closed, which is too late to trade on at yesterday's closing price.
- **Chance**: three years is not long. Good results can simply be luck.

A fair test is to choose your words using one period (say, 2023 to 2024), then check them **once** on a period you haven't looked at (2025).
