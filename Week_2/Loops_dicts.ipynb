## Dictionaries

A **`dict`** is a mutable, unordered collection of key–value pairs where keys are unique and values can be anything. Lookup time in a `dict` is `O(1)` on average, when in a list it's `O(n)`: this means it, for a `dict` of any size, it takes the <i>same</i> time to look up an item, where in a `list` the lookup time scales with the size of the `list`. This makes the `dict` the better structure for any process where speed is important.

To create a `dict` in python, you use curly braces and separate the key and the value with a colon `:` as follows:
```python
    animals = {'dog': 'mammal', 'spider':'arachnid', 'hornet':'insect', 'toad':'amphibian'}
```
To access a value from a `dict`, you put the key in square brackets after the `dict` name:
```python
    animals['hornet'] = 'insect'
```
Note that a `dict` can contain a `dict` or another data container like a `list`:

```python
edinburgh = {"name": "Edinburgh", "population": 506520, "postcodes": ["EH1", "EH8", "EH16"], "landmarks": ["Edinburgh Castle", "Arthur's Seat"], "council": {"name": "City of Edinburgh Council", "headquarters": "Waverley Court"},"districts": {"north": ["Leith", "Newhaven"], "south": ["Morningside","Bruntsfield"]}}
```
Values can be accessed by blending key-based access and indexing:
```python
    edinburgh["postcodes"] = ['EH1', 'EH8', 'EH16']  # dict key gives a list
    edinburgh["postcodes"][0] = 'EH1'                     # dict key and list index gives string value
    edinburgh["districts"]["north"] = ['Leith', 'Newhaven']     # dict key and dict key gives list
    edinburgh["districts"]["north"][1] = 'Newhaven'  # dict key and dict key and list index gives string value
```
Note that **keys must be unique, but values may repeat**. This makes a dict a natural fit for a **many-to-one mapping**, which is a very common structure in real-world data.

## Iteration

A **`for` loop** repeats a block of code once for each item in an **iterable** — a container or sequence you can step through, such as a `list`, `tuple`, `str`, `dict`, `set`, or `range`. Each pass through the loop is called an **iteration**, and the current item is bound to a variable you name in the loop header. Unlike a `while` loop, you don't manage a counter or a stop condition yourself: the loop walks the iterable from start to finish and stops automatically when the items run out.

To write a `for` loop in Python, you use the keyword `for`, a variable name for the current item, the keyword `in`, the iterable, and a colon `:`. The body is indented beneath it:

```python
    animals = ['dog', 'spider', 'hornet', 'toad']

    for animal in animals:
        print(animal)
```

The variable name (`animal`) is yours to choose — `for x in animals:` works just as well. The loop runs four times, once per item, and then ends. Note that the loop variable keeps its **last** value after the loop finishes:

```python
    print(animal)   # 'toad'
```

### Looping over a `range`

When you want to repeat something a fixed number of times, or step through numbers, use `range()`. It produces a sequence of integers on demand:

```python
    for i in range(5):        # 0, 1, 2, 3, 4
        print(i)

    for i in range(1, 4):     # 1, 2, 3  (start is inclusive, stop is exclusive)
        print(i)

    for i in range(0, 10, 2): # 0, 2, 4, 6, 8  (third value is the step)
        print(i)
```

### Looping over a `dict`

A `dict` is iterable, but there are three distinct things you might want to loop over: its **keys**, its **values**, or its **key–value pairs**. Python gives you a method for each:

```python
    animals = {'dog': 'mammal', 'spider': 'arachnid', 'hornet': 'insect', 'toad': 'amphibian'}

    for key in animals:                 # loops over keys by default
        print(key)                      # 'dog', 'spider', ...

    for value in animals.values():      # loops over values
        print(value)                    # 'mammal', 'arachnid', ...

    for key, value in animals.items():  # loops over key-value pairs
        print(key, '->', value)         # 'dog -> mammal', ...


````python
animals = {'dog': 'mammal', 'spider': 'arachnid', 'hornet': 'insect', 'toad': 'amphibian'}

for key, value in animals.items():         # loops over keys by default
    print(key, value)                      # 'dog', 'spider', ...
````

````python
placename_elements = {
    "Brittonic": {
        "prefix": ["aber", "caer", "llan", "lan", "pen", "tre", "tref", "pol", "cwm", "coed", "nant"],
        "suffix": [],
    },
    "Goidelic": {
        "prefix": ["bal", "kil", "inver", "strath", "glen", "ard"],
        "suffix": [],
    },
    "Latin": {
        "prefix": ["strat", "stret"],
        "suffix": ["chester", "cester", "caster"],
    },
    "Old English": {
        "prefix": ["chipping", "stoke"],
        "suffix": ["ham", "ton", "ley", "leigh", "worth", "field", "ford", "bury", "borough", "stead", "hurst", "den"],
    },
    "Norse": {
        "prefix": ["kirk"],
        "suffix": ["by", "thorpe", "thorp", "thwaite", "toft", "holme", "beck", "gill", "ay"],
    }
}
````

````python
import pandas as pd
import matplotlib.pyplot as plt

# Empty lists to fill as we go
names = []
origins = []
eastings = []
northings = []

# Looping over a dictionary gives us its keys: here, each place name
for name in places:
    info = places[name]          # look up that place's details, e.g. places["Oxford"]
    lower = name.lower()
    origin = "Unclassified"      # default, in case nothing matches

    # Loop over the languages: "Brittonic", "Goidelic", "Latin", ...
    for language in placename_elements:
        elements = placename_elements[language]      # that language's prefixes and suffixes

        for prefix in elements["prefix"]:
            if lower.startswith(prefix):
                origin = language

        for suffix in elements["suffix"]:
            if lower.endswith(suffix):
                origin = language

    # Store the result for this place
    names.append(name)
    origins.append(origin)
    eastings.append(info["easting"])
    northings.append(info["northing"])

df = pd.DataFrame({"name": names, "origin": origins,
                   "easting": eastings, "northing": northings})
df.head()
````

````python
# 2. Plot: the same structure as the pub map
fig, ax = plt.subplots(figsize=(8, 12))
ax.scatter(df.easting, df.northing, s=0.3, c="black", alpha=0.4, linewidths=0)

for origin, colour in [("Norse", "tab:red")]:
    subset = df[df.origin == origin]
    ax.scatter(subset.easting, subset.northing, s=6, c=colour, label=origin)

ax.set_aspect("equal")
ax.axis("off")
ax.legend(markerscale=3)
plt.title("British placenames by\n linguistic origin")
plt.show()
````
